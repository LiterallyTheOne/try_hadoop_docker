## Code

```python
data_path = Path("my_data/youtube.txt")
```

```python
whole_data = []

with data_path.open("r") as d1:
    data_content = d1.readlines()
    for x in data_content:
        x = x.replace("\n", "")
        whole_data.append(x.split("\t"))
```
